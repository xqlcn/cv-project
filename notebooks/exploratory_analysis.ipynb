# Exploratory analysis

**ModelNet40:** scan `data/modelnet40/train/` and inspect the **first five categories** (alphabetically) — mesh paths, counts, quick table.

**Optional later:** `data/metadata/chirality_index.jsonl` (Blender) and `data/features/*/features_index.jsonl` for probe runs.

```python
from pathlib import Path
import sys

import pandas as pd

# Notebook lives in notebooks/ — project root is parent
PROJECT = Path("..").resolve()
if str(PROJECT) not in sys.path:
    sys.path.insert(0, str(PROJECT))

from src.datasets.modelnet40_dataset import discover_modelnet40_records

mn_root = PROJECT / "data" / "modelnet40"
train_records = discover_modelnet40_records(mn_root, splits=("train",))

# First 5 categories alphabetically
all_categories = sorted({r["category"] for r in train_records})
selected = all_categories[:5]
rows = [r for r in train_records if r["category"] in selected]

df = pd.DataFrame(rows)
print(f"Root: {mn_root.resolve()}")
print(f"Using categories {selected} (5 of {len(all_categories)} total)")
print(f"Rows in subset: {len(df)}")
df.head(12)
```

```python
# Per-category mesh counts (same 5 categories)
df.groupby("category").size().reindex(selected).to_frame("n_train_meshes")
```
